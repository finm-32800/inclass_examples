# Demo: Package-Anchored Paths

This notebook demonstrates that the same imports work identically in notebooks as they do in scripts.

**Key point**: No `os.chdir()` hacks needed. The path is anchored on the installed package location.

````python
from myproj import project_root, data_dir, data_path, output_dir, output_path

print(f"Project root:     {project_root()}")
print(f"Data directory:   {data_dir()}")
print(f"Output directory: {output_dir()}")
````

## Why This Works

The `myproj` package is installed in editable mode (`pip install -e .`). This means:

1. Python can find `myproj` from anywhere
2. `myproj.__file__` points to the actual source file
3. We calculate project root relative to that file location

The calculation in `paths.py`:
```python
Path(__file__).resolve().parents[2]
#    paths.py -> myproj/ -> src/ -> project_root/
#    That's 3 levels up, so parents[2] (0-indexed)
```

````python
# Let's demonstrate by creating a file in _data/ and reading it back

# Ensure _data directory exists
data_dir().mkdir(exist_ok=True)

# Write a sample file
sample_file = data_path("sample.txt")
sample_file.write_text("Hello from the notebook!\nThis file was created using package-anchored paths.")
print(f"Created: {sample_file}")
````

````python
# Read it back to verify
content = sample_file.read_text()
print("File contents:")
print(content)
````

## Comparison with Other Approaches

| Approach | Pros | Cons |
|----------|------|------|
| **Package (this)** | Clean imports, works everywhere | Requires `pip install -e .` |
| Marker walk | No install needed | Slower, more code |
| `os.chdir()` | Simple | Fragile, notebook-specific |
| Environment vars | Flexible | Must manage env |

The package approach is ideal when you're already using virtual environments (which you should be!).
